```python
# one row per track_id, keeping the highest popularity, and collect all genres
genres = df.groupby("track_id")["track_genre"].agg(lambda s: ", ".join(sorted(set(s))))
df = (df.sort_values("popularity", ascending=False)
        .drop_duplicates(subset="track_id")
        .reset_index(drop=True))
df["all_genres"] = df["track_id"].map(genres)

# same song re-released under different IDs (singles vs albums, remasters)
df["_title"] = (df["track_name"].str.lower()
                  .str.replace(r"\(.*?\)|\[.*?\]", "", regex=True)
                  .str.replace(r"\s+", " ", regex=True).str.strip())
df["_artist"] = df["artists"].str.lower().str.strip()
df = df.drop_duplicates(subset=["_title", "_artist"]).reset_index(drop=True)

df.to_csv("data/spotify_tracks_clean.csv", index=False)
```

```python
features = ["danceability", "energy", "acousticness", "instrumentalness",
            "liveness", "speechiness", "valence", "tempo", "loudness",
            "key_sin", "key_cos", "mode"]

print(df[features].describe().T)
print("tempo == 0:", (df["tempo"] == 0).sum())
print("time_signature == 0:", (df["time_signature"] == 0).sum())
print("key == -1:", (df["key"] == -1).sum())
print("popularity == 0:", (df["popularity"] == 0).sum())
print("duration < 30s or > 15min:",
      ((df["duration_ms"] < 30_000) | (df["duration_ms"] > 900_000)).sum())
```

```python
before = len(df)

bad = (
    (df["tempo"] == 0)
    | (df["time_signature"] == 0)
    | (df["duration_ms"] < 30_000)
    | (df["duration_ms"] > 900_000)
    | (df["loudness"] < -35)
)
print("rows to drop:", bad.sum())   # check before dropping
df = df[~bad].reset_index(drop=True)
print(before, "->", len(df))

df.to_csv("data/spotify_tracks_clean.csv", index=False)
```
